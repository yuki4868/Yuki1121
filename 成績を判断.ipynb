```python
num = int(input("整数を入力してください"))

if num >= 90:
  print("Sです。合格です。")
elif 90 > num >= 80:
  print("Aです。合格です。")
elif 80 > num >= 70:
  print("Bです。合格です。")
elif 70 > num >= 60:
  print("Cです。合格です。")
else:
  print("不合格です。")
```
